```python
import pandas as pd

df = pd.read_csv('Desktop/DA/sales project 1/customer_shopping_behavior.csv')

print("Dataset loaded successfully!")
```

```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(
    lambda x: x.fillna(x.median())
)
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ', '_')
df = df.rename(columns={'purchase_amount_(usd)': 'purchase_amount'})
```

```python
df.columns
```

```python
# Create an age group column

labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']

df['age_group'] = pd.cut(df['age'], q=4, labels=labels)
```

```python
# Create an age group column

labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']

df['age_group'] = pd.qcut(df['age'], q=4, labels=labels)
```

```python
df[['age', 'age_group']].head(10)
```

```python
# Create column purchase_frequency_days

frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase_frequency_days', 'frequency_of_purchases']].head(10)
```

```python
from sqlalchemy import create_engine
from sqlalchemy.engine import URL

# PostgreSQL connection details
username = "postgres"
password = "PgAdmin@2026#Data!"  # Enter your actual PostgreSQL password
host = "localhost"
port = 5432
database = "customer_behavior"

# Create a safe connection URL
connection_url = URL.create(
    drivername="postgresql+psycopg2",
    username=username,
    password=password,
    host=host,
    port=port,
    database=database
)

engine = create_engine(connection_url)

# Load DataFrame into PostgreSQL
table_name = "customer"

df.to_sql(table_name, engine, if_exists="replace", index=False)

print("Data successfully loaded into PostgreSQL!")
```
